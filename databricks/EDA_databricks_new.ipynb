<p style="font-size: 18px; color: #f1f1f1; background: #2c3e50; padding: 15px; border-radius: 10px;">
    📥 <strong>Resolve Missing Values and Duplicates:</strong>
</p>

<p style="font-size: 16px; color: #7f8c8d; padding: 20px; border-radius: 10px;">
    <strong><code>Close_NVDA</code></strong>, <strong><code>High_NVDA</code></strong> and <strong><code>Low_NVDA</code></strong>:
</p>
<p style="font-size: 16px; color: #7f8c8d; margin-left: 30px; border-radius: 10px;">
    We use this approach because stock prices follow a <strong>sequential trend</strong>, meaning that each day's price is influenced by the previous day's data. If a trading day's data is missing, the <strong> last known price is the best estimate</strong> until new values appear. This prevents <strong>unrealistic jumps</strong> in stock prices caused by missing data.
    Additionally, since stock trading is <strong>continuous</strong>, gaps in the data can occur because stock exchanges <strong>do not operate on weekends or holidays</strong>. As a result, missing values may appear. Using forward fill ensures that the <strong>dataset remains consistent</strong> and prevents artificial dips in stock prices.
    Finally, this method <strong>preserves data consistency for analysis and modelling</strong>, ensuring that financial calculations remain smooth and uninterrupted.
    <br>
    To implement a forward-fill strategy we used PySpark window functions. This approach relies on defining a time-ordered window over the dataset (sorted by <code>Date</code>) and using the last non-null value observed up to the current row.
    <br>
    Specifically, for each column, we applied the <code>last(..., ignorenulls=True)</code> function within a window that spans from the beginning of the dataset up to the current date. This ensures that any missing value is replaced by the most recent known value chronologically.
    <br><br>
</p>

```python
# Define window spec ordered by date
# For each row, the window includes all rows from the beginning of time up to that date, and orders it by date 
window_spec = Window.partitionBy().orderBy("Date").rowsBetween(Window.unboundedPreceding, 0) 

# Forward fill each column
# Updates each column by looking backward from the current row and filling any missing value with the most recent known value
df = df.withColumn("Close_NVDA", last("Close_NVDA", ignorenulls=True).over(window_spec)) \
       .withColumn("High_NVDA", last("High_NVDA", ignorenulls=True).over(window_spec)) \
       .withColumn("Low_NVDA", last("Low_NVDA", ignorenulls=True).over(window_spec))
```

<p style="font-size: 16px; color: #7f8c8d; padding: 20px; border-radius: 10px;">
    <strong><code>Market_Cap_NVDA</code></strong> and <strong><code>EV_NVDA</code></strong>
</p>

<p style="font-size: 16px; color: #7f8c8d; margin-left: 30px; border-radius: 10px;">
    For the variables <strong><code>Market_Cap_NVDA</code></strong> and <strong><code>EV_NVDA</code></strong>, we apply the approach of <strong>replacing missing values with zero (0)</strong> because these financial metrics are derived from calculations that may not always be available in certain periods.
</p>

<p style="font-size: 16px; color: #7f8c8d; margin-left: 40px; border-radius: 10px;">
    📌 <strong>Market Capitalization</strong> is calculated as: <code>Market_Cap_NVDA = Close_NVDA × Shares_Out_NVDA</code>

<p style="font-size: 16px; color: #7f8c8d; margin-left: 40px; border-radius: 10px;">

</p>

<p style="font-size: 16px; color: #7f8c8d; margin-left: 30px; border-radius: 10px;">
    If any of these components are missing, it means that there is insufficient data to compute a meaningful value. Instead of keeping missing values (<code>NaN</code>), we assign <strong>zero (0)</strong> to prevent issues in further analysis, particularly when performing mathematical operations, where missing values could affect statistical computations and financial models. Additionally, this approach <strong>avoids distortions</strong> in charts and data visualizations, ensuring that calculations remain consistent over time.
</p>

```python
df = df.fillna({'Market_Cap_NVDA': 0, 'EV_NVDA': 0})
```

<p style="font-size: 16px; color: #7f8c8d; padding: 20px; border-radius: 10px;">
    <strong><code>Total_Debt_NVDA</code></strong> and <strong><code>Cash_Equivalents_NVDA</code></strong>
</p>

<p style="font-size: 16px; color: #7f8c8d; margin-left: 30px; border-radius: 10px;">
    For the variables <strong><code>Total_Debt_NVDA</code></strong> and <strong><code>Cash_Equivalents_NVDA</code></strong>, we handle missing values by filling them with the <strong>median</strong> of the dataset instead of using forward-fill or replacing them with zero.
</p>

<p style="font-size: 16px; color: #7f8c8d; margin-left: 40px; border-radius: 10px;">
    📌 <strong>Total Debt</strong> represents the sum of short-term and long-term liabilities that a company holds.<br>
    📌 <strong>Cash Equivalents</strong> refer to the company's liquid assets that can be quickly converted into cash.
</p>

<p style="font-size: 16px; color: #7f8c8d; margin-left: 30px; border-radius: 10px;">
    Using the median instead of zero helps maintain a more <strong>realistic financial representation</strong> by preventing extreme fluctuations that could result from missing values. Since financial liabilities and cash reserves usually follow a stable trend, the median is a better estimate than the mean, as it reduces the impact of outliers and ensures more <strong>robust statistical imputation</strong>. This approach ensures that missing values do not negatively affect financial calculations such as <code>Enterprise Value (EV_NVDA)</code> and other investment analysis metrics.
</p>

```python
# Replace 0 with null 
df = df.withColumn("Total_Debt_NVDA", when(col("Total_Debt_NVDA") == 0, None).otherwise(col("Total_Debt_NVDA")))
df = df.withColumn("Cash_Equivalents_NVDA", when(col("Cash_Equivalents_NVDA") == 0, None).otherwise(col("Cash_Equivalents_NVDA")))

# Cast strings to float to avoid errors
df = df.withColumn("Total_Debt_NVDA", col("Total_Debt_NVDA").cast("double"))
df = df.withColumn("Cash_Equivalents_NVDA", col("Cash_Equivalents_NVDA").cast("double"))

# Compute median (approximate)
medians = df.select(
    percentile_approx("Total_Debt_NVDA", 0.5).alias("med_TD"),
    percentile_approx("Cash_Equivalents_NVDA", 0.5).alias("med_CE")
).collect()[0]

# Fill nulls with median
df = df.fillna({
    "Total_Debt_NVDA": medians["med_TD"],
    "Cash_Equivalents_NVDA": medians["med_CE"]
})
```

<p style="font-size: 16px; color: #7f8c8d; padding: 20px; border-radius: 10px;">
    <strong><code>Volume_NVDA</code></strong> and <strong><code>Shares_Out_NVDA</code></strong>
</p>

<p style="font-size: 16px; color: #7f8c8d; margin-left: 30px; border-radius: 10px;">
    Since <code>Volume_NVDA</code> and <code>Shares_Out_NVDA</code> both represent financial data that generally follows a continuous trend (one point comes after another), we first tried to apply <strong>linear interpolation</strong> to substitute missed values.  This technique fills in missed values by calculating intermediate values between known data points providing a more seamless and accurate forecast while respecting stock market trends. 
    However, Native PySpark does not support linear interpolation directly. Therefore, we tried to use the <code>pyspark.pandas</code> API (formerly Koalas), which provides a pandas-like interface on top of Spark. This should allow us to perform linear interpolation at scale using familiar pandas syntax while still leveraging the distributed nature of Spark.
    This would be ideal because, by invoking <code>interpolate(method='linear')</code>, we ensure that missed values are filled with estimates derived from immediately neighbouring data points, in consideration for maintaining the dataset's continuity to avoid any unwanted spikes in the stock data.
    However, transforming the df to a pandas-on-Spark and then re-converting it caused problems later on, as it messed with the Date variable. Therefore, while you can still see our initial approach commented below, we ultimately decided to go for the <code>forward fill</code> strategy based on PySpark window functions. This method ensures that each missing value is replaced with the most recent known value in chronological order, maintaining continuity without introducing artificial jumps.
    This forward-fill approach is computationally efficient and preserves temporal consistency.

```python
from pyspark.sql.functions import mean

mean_vals = df.select([mean(c).alias(c) for c in ['Volume_NVDA', 'Shares_Out_NVDA']]).collect()[0]

for colname in ['Volume_NVDA', 'Shares_Out_NVDA']:
    df = df.fillna({colname: mean_vals[colname]})

# After the resolve of nan values we recalculate Market Cap
df = df.withColumn("Market_Cap_NVDA", col("Close_NVDA") * col("Shares_Out_NVDA"))

# Recalculate Price-to-Sales Ratio
df = df.withColumn("PS_Ratio_NVDA", col("Market_Cap_NVDA") / col("Total_Revenue_NVDA"))

# Handle missing or infinite values
df = df.withColumn(
    "PS_Ratio_NVDA",
    when((col("PS_Ratio_NVDA").isNull()) | (col("PS_Ratio_NVDA") == float("inf")) | (col("PS_Ratio_NVDA") == float("-inf")), 0)
    .otherwise(col("PS_Ratio_NVDA"))
)



```

    Analyzing the daily percentage change in stock price is important for assessing volatility and short-term changes in investor attitude. Understanding the difference in the closing price on a day-to-day basis gives indications of extreme high volatility, sudden market movements, or even a smoother change. Such information can be impactful in defining risk and guiding forecasting models to provide more accurate, data-driven predictions.

```python
# Define window ordered by Date
w = Window.orderBy("Date")

# Compute percentage change
df = df.withColumn(
    "Daily_Change_%",
    when(
        lag("Close_NVDA").over(w).isNotNull(),
        ((col("Close_NVDA") - lag("Close_NVDA").over(w)) / lag("Close_NVDA").over(w)) * 100
    ).otherwise(0.0)
)
```

<p style="font-size: 16px; color: #7f8c8d; padding: 20px; border-radius: 10px;">
    In order to define unusual trading volume, we employ a dynamic threshold based on the statistical distribution of each stock volume's data. Specifically, we define <code>unusual activity</code> as any volume which exceeds <strong>the average volume, plus two standard deviations</strong>:
    <br><br>
    <code style="color: #1abc9c; background-color: #2c3e50; padding: 3px 8px; border-radius: 5px;">threshold = mean + 2 × std</code>
</p>

<p style="font-size: 16px; color: #7f8c8d; padding: 10px 20px; border-radius: 10px;">
    This is based on statistical reasoning: in a normal distribution, approximately 95% of the data is found within two standard deviations of the mean. By choosing this as our threshold, we will identify only the top 5% of extreme values as <strong>unusual</strong>, the true outliers.
</p>

<p style="font-size: 16px; color: #7f8c8d; padding: 10px 20px; border-radius: 10px;">
    Spikes in trading activity generally indicate important market events like earnings releases, geopolitical news or surprise countervailing buyer or seller reactions.  Spikes will allow you to identify periods of heightened interest and possible price changes. For this reason, they are a good indicator for short-term traders as well as long-term analysts.
</p>>

```python
# Compute mean and stddev of Volume_NVDA
stats = df.select(
    mean("Volume_NVDA").alias("avg_volume"),
    stddev("Volume_NVDA").alias("std_volume")
).collect()[0]

# Calculate threshold
unusual_threshold = stats["avg_volume"] + 2 * stats["std_volume"]

# Add Unusual_Activity column
df = df.withColumn("Unusual_Activity", col("Volume_NVDA") > unusual_threshold)
```

```python
def volume_act_analysis(df):
    fig = go.Figure()

    fig.add_trace(go.Scatter(
        x=df['Date'],
        y=df['Volume_NVDA'],
        mode='lines',
        name='Volume',
        line=dict(color='#1abc9c', width=2.2),
        hovertemplate='📅 %{x}<br>📦 Volume: %{y}<extra></extra>'
    ))

    fig.add_trace(go.Scatter(
        x=df[df['Unusual_Activity']]['Date'],
        y=df[df['Unusual_Activity']]['Volume_NVDA'],
        mode='markers',
        name='Unusual Activity',
        marker=dict(color='#e74c3c', size=8, symbol='circle'),
        hovertemplate='📅 %{x}<br>🚨 Unusual Volume: %{y}<extra></extra>'
    ))

    fig.update_layout(
        updatemenus=[dict(
            type="buttons",
            direction="right",
            x=0,
            xanchor="left",
            y=1.15,
            yanchor="top",
            bgcolor="#34495e",
            bordercolor="#7f8c8d",
            borderwidth=1,
            font=dict(color="white", size=13),
            buttons=[
                dict(method="relayout", label="1Y",
                     args=[{"xaxis.range": [df['Date'].max() - pd.DateOffset(years=1), df['Date'].max()],
                            "title": "📊 NVIDIA – Volume Activity (1 Year)"}]),
                dict(method="relayout", label="5Y",
                     args=[{"xaxis.range": [df['Date'].max() - pd.DateOffset(years=5), df['Date'].max()],
                            "title": "📊 NVIDIA – Volume Activity (5 Years)"}]),
                dict(method="relayout", label="10Y",
                     args=[{"xaxis.range": [df['Date'].max() - pd.DateOffset(years=10), df['Date'].max()],
                            "title": "📊 NVIDIA – Volume Activity (10 Years)"}]),
                dict(method="relayout", label="All",
                     args=[{"xaxis.autorange": True,
                            "title": "📊 NVIDIA – Volume and Market Activity"}])
            ]
        )]
    )

    fig.update_layout(
        title=dict(
            text="📊 NVIDIA – Volume and Market Activity",
            font=dict(size=20, color="white"),
            x=0.5
        ),
        xaxis=dict(
            title="Date",
            showgrid=True,
            gridcolor="#34495e",
            linecolor="#7f8c8d",
            rangeslider=dict(visible=True, thickness=0.04),
            type="date"
        ),
        yaxis=dict(
            title="Trading Volume",
            showgrid=True,
            gridcolor="#34495e"
        ),
        plot_bgcolor="#1e272e",
        paper_bgcolor="#1e272e",
        font=dict(color="white"),
        hovermode="x unified",
        height=650,
        margin=dict(t=90, b=60, l=60, r=60)
    )

    fig.show()


```
