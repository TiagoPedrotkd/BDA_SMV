```python
def processar_csv_e_calcular_media_pyspark(pasta_dbfs_path):
    lista_dfs = []

    print(f"Buscando arquivos CSV na pasta: {pasta_dbfs_path}")
    arquivos = dbutils.fs.ls(pasta_dbfs_path)
    for arquivo in arquivos:
        if arquivo.name.endswith(".csv"):
            caminho = arquivo.path
            print(f"  - Encontrado: {arquivo.name}")
            try:
                df = spark.read.option("header", True).csv(caminho)
                lista_dfs.append(df)
            except Exception as e:
                print(f"    ! Erro ao ler {arquivo.name}: {e}")

    if not lista_dfs:
        print(f"Nenhum arquivo CSV encontrado na pasta: {pasta_dbfs_path}")
        return None

    df_completo = lista_dfs[0]
    for df in lista_dfs[1:]:
        df_completo = df_completo.unionByName(df, allowMissingColumns=True)

    print("DataFrames concatenados com sucesso.")
    df_completo.show(5)

    coluna_chave = df_completo.columns[0]
    print(f"Assumindo '{coluna_chave}' como coluna chave para agrupar.")

    tipos = dict(df_completo.dtypes)
    colunas_numericas = [c for c,t in tipos.items() if t in ('int', 'bigint', 'double', 'float')]
    colunas_nao_numericas = [c for c in df_completo.columns if c not in colunas_numericas]

    print(f"Colunas numéricas: {colunas_numericas}")
    print(f"Colunas não numéricas: {colunas_nao_numericas}")

    agg_exprs = []

    for c in colunas_numericas:
        if c != coluna_chave:
            agg_exprs.append(sum(col(c)).alias(c))

    for c in colunas_nao_numericas:
        if c != coluna_chave:
            agg_exprs.append(first(col(c), ignorenulls=True).alias(c))

    df_agrupado = df_completo.groupBy(coluna_chave).agg(*agg_exprs)

    contagens = df_completo.groupBy(coluna_chave).count()

    df_agrupado = df_agrupado.join(contagens, on=coluna_chave)

    for c in colunas_numericas:
        if c != coluna_chave and c in df_agrupado.columns:
            df_agrupado = df_agrupado.withColumn(c, col(c) / col("count"))

    df_agrupado = df_agrupado.drop("count")

    print("Agrupamento e média calculados com sucesso.")
    df_agrupado.show(5)

    return df_agrupado
```

```python
from sklearn.preprocessing import MinMaxScaler

def normalizar_dados(df, colunas_features, colunas_target):
    for col in colunas_features + colunas_target:
        df[col] = pd.to_numeric(df[col], errors='coerce')
    df = df.dropna(subset=colunas_features + colunas_target)
    
    df_scaled = df.copy()
    
    scaler_features = MinMaxScaler(feature_range=(0,1))
    df_scaled[colunas_features] = scaler_features.fit_transform(df_scaled[colunas_features])
    
    scaler_target = MinMaxScaler(feature_range=(0,1))
    df_scaled[colunas_target] = scaler_target.fit_transform(df_scaled[colunas_target])
    
    return df_scaled, scaler_features, scaler_target
```

```python
columns_numeric = ['high','low','RSI_rsi','SMA_sma','EMA_ema','MACD_macd','MACD_macd_signal','MACD_macd_hist','STOCH_slow_k','STOCH_slow_d','ADX_adx','CCI_cci','ATR_atr','BBANDS_upper_band','BBANDS_middle_band','BBANDS_lower_band']

columns_target = ['close', 'open', 'volume']

df_scaled, scaler_features, scaler_target = normalizar_dados(df_pd, columns_numeric, columns_target)
```

```python
def preparar_dados_prophet(df, coluna_data, coluna_hora, coluna_valor):
    df[coluna_data] = df[coluna_data].astype(str).str.strip()
    df[coluna_hora] = df[coluna_hora].astype(str).str.strip()

    df['ds'] = pd.to_datetime(df[coluna_data] + ' ' + df[coluna_hora], errors='coerce')

    num_nat = df['ds'].isna().sum()
    if num_nat > 0:
        print(f"AVISO: {num_nat} valores NaT (data/hora inválidas) foram gerados.")

    df_prophet = df[['ds', coluna_valor]].rename(columns={coluna_valor: 'y'})

    df_prophet = df_prophet.dropna(subset=['ds'])

    return df_prophet
```

```python
def criar_sequencias_multivariadas_multitarget(df_scaled, colunas_features, colunas_alvo, n_steps):
    import itertools
    if any(isinstance(el, list) for el in colunas_features):
        colunas_features = list(itertools.chain.from_iterable(colunas_features))
    if any(isinstance(el, list) for el in colunas_alvo):
        colunas_alvo = list(itertools.chain.from_iterable(colunas_alvo))
    
    df_filtered = df_scaled.dropna(subset=colunas_features + colunas_alvo)
    
    data_features = df_filtered[colunas_features].values
    data_targets = df_filtered[colunas_alvo].values
    
    X, y = [], []
    
    for i in range(len(df_filtered) - n_steps):
        seq_x = data_features[i:i + n_steps]
        seq_y = data_targets[i + n_steps]
        X.append(seq_x)
        y.append(seq_y)
    
    X = np.array(X)
    y = np.array(y)
    
    return X, y
```

```python
n_steps = 60
X, y = criar_sequencias_multivariadas_multitarget(df_scaled, columns_numeric, columns_target, n_steps)
```

```python
def dividir_treino_teste(X, y, treino_ratio=0.8):
    tamanho_treino = int(len(X) * treino_ratio)
    
    X_train = X[:tamanho_treino]
    y_train = y[:tamanho_treino]
    X_test = X[tamanho_treino:]
    y_test = y[tamanho_treino:]
    
    return X_train, y_train, X_test, y_test
```

```python
def criar_modelo_multitarget_simple_rnn(n_steps, n_features, n_targets):
    model = Sequential([
        SimpleRNN(50, return_sequences=True, input_shape=(n_steps, n_features)),
        SimpleRNN(50),
        Dense(n_targets)
    ])
    model.compile(optimizer='adam', loss='mse', metrics=['mae'])
    return model
```

```python
n_features = len(columns_numeric)
n_targets = len(columns_target)

model = criar_modelo_multitarget_simple_rnn(n_steps, n_features, n_targets)
```

```python
def treinar_modelo(model, X_train, y_train, epochs=50, batch_size=32):
    from tensorflow.keras.callbacks import EarlyStopping

    early_stop = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
    history = model.fit(
        X_train, y_train,
        epochs=epochs,
        batch_size=batch_size,
        validation_split=0.2,
        callbacks=[early_stop]
    )
    return history
```
