<p style="font-size: 16px; color: #7f8c8d; padding: 15px;">
    <strong><code>close</code>, <code>high</code>, <code>low</code>, <code>open</code>:</strong>
</p>
<p style="font-size: 16px; color: #7f8c8d; margin-left: 30px;">
    Forward-fill is used to handle missing values in price columns. Since stock prices follow a <strong>sequential trend</strong>, using the last available value ensures continuity and avoids price jumps. This is particularly relevant on weekends or holidays when markets are closed.
</p>

```python
window_spec = Window.orderBy("Date").rowsBetween(Window.unboundedPreceding, 0)

df = df.withColumn("Close_NVDA", last("Close_NVDA", ignorenulls=True).over(window_spec)) \
       .withColumn("High_NVDA", last("High_NVDA", ignorenulls=True).over(window_spec)) \
       .withColumn("Low_NVDA", last("Low_NVDA", ignorenulls=True).over(window_spec))
```

```python
window_spec = Window.orderBy("Date").rowsBetween(-sys.maxsize, 0)

df = df.withColumn("Volume_NVDA", last("Volume_NVDA", ignorenulls=True).over(window_spec))
```

```python
def time_series_analysis(df_daily: pd.DataFrame, df_full: pd.DataFrame):
    fig = go.Figure()

    fig.add_trace(go.Scatter(
        x=df_daily['Date'],
        y=df_daily['Close_NVDA'],
        mode='lines',
        name='Daily',
        line=dict(color='#3498db', width=2.2, shape='spline', smoothing=1.2),
        visible=True
    ))

    max_date = df_full['Date'].max()

    fig.update_layout(
        updatemenus=[
            dict(
                type="buttons",
                direction="right",
                x=0,
                xanchor="left",
                y=1.15,
                yanchor="top",
                bgcolor="#34495e",
                bordercolor="#7f8c8d",
                borderwidth=1,
                font=dict(color="white", size=13),
                buttons=[
                    dict(method="relayout", label="1Y",
                         args=[{"xaxis.range": [max_date - pd.DateOffset(years=1), max_date],
                                "title": "📈 NVIDIA Closing Prices - Last 1 Year"}]),
                    dict(method="relayout", label="5Y",
                         args=[{"xaxis.range": [max_date - pd.DateOffset(years=5), max_date],
                                "title": "📈 NVIDIA Closing Prices - Last 5 Years"}]),
                    dict(method="relayout", label="10Y",
                         args=[{"xaxis.range": [max_date - pd.DateOffset(years=10), max_date],
                                "title": "📈 NVIDIA Closing Prices - Last 10 Years"}]),
                    dict(method="relayout", label="All",
                         args=[{"xaxis.autorange": True,
                                "title": "📈 NVIDIA Closing Prices Over Time"}])
                ]
            )
        ]
    )

    fig.update_layout(
        xaxis=dict(
            rangeslider=dict(visible=True, thickness=0.04),
            type="date",
            showline=True,
            linecolor="#7f8c8d"
        ),
        title=dict(
            text="📈 NVIDIA Closing Prices Over Time",
            font=dict(size=20, color="white"),
            x=0.5
        ),
        yaxis_title="Closing Price (USD)",
        plot_bgcolor="#1e272e",
        paper_bgcolor="#1e272e",
        font=dict(color="white"),
        hovermode="x unified",
        margin=dict(t=110, b=60, l=60, r=60),
        height=700
    )

    fig.show()
```

```python
df_pd = df.toPandas()
df_pd["Date"] = pd.to_datetime(df_pd["Date"]) 
df_pd['Close_NVDA'] = pd.to_numeric(df_pd['Close_NVDA'], errors='coerce')
df_pd['Open_NVDA'] = pd.to_numeric(df_pd['Open_NVDA'], errors='coerce')
df_pd['High_NVDA'] = pd.to_numeric(df_pd['High_NVDA'], errors='coerce')
df_pd['Low_NVDA'] = pd.to_numeric(df_pd['Low_NVDA'], errors='coerce')
df_pd['Volume_NVDA'] = pd.to_numeric(df_pd['Volume_NVDA'], errors='coerce')

time_series_analysis(df_pd, df_pd)
```

<p style="font-size: 18px; color: #f1f1f1; background: #2c3e50; padding: 15px; border-radius: 10px;">
    📥 <strong>Daily Percentage Change:</strong>
</p>

<p style="font-size: 16px; color: #7f8c8d; padding: 20px; border-radius: 10px;">
    Analyzing the daily percentage change in stock price is important for assessing volatility and short-term changes in investor attitude. Understanding the difference in the closing price on a day-to-day basis gives indications of extreme high volatility, sudden market movements, or even a smoother change. Such information can be impactful in defining risk and guiding forecasting models to provide more accurate, data-driven predictions.
</p>


```python
window_spec = Window.orderBy("Date")

df = df.withColumn(
    "Daily_Change",
    when(
        lag("Close_NVDA").over(window_spec).isNotNull(),
        ((col("Close_NVDA") - lag("Close_NVDA").over(window_spec)) / lag("Close_NVDA").over(window_spec))
    ).otherwise(0.0)
)
```

```python
def daily_change_analysis(df_pd: pd.DataFrame):
    fig = go.Figure()

    fig.add_trace(go.Scatter(
        x=df_pd['Date'],
        y=df_pd['Daily_Change'],
        mode='lines',
        name='Daily Change / Return',
        line=dict(color='#f39c12', width=2.2, shape='spline', smoothing=1.2),
        hovertemplate='📅 %{x}<br>📈 Change: %{y:.2f}%<extra></extra>'
    ))

    max_date = df_pd['Date'].max()
    fig.update_layout(
        updatemenus=[
            dict(
                type="buttons",
                direction="right",
                x=0,
                xanchor="left",
                y=1.18,
                yanchor="top",
                bgcolor="#34495e",
                bordercolor="#7f8c8d",
                borderwidth=1,
                font=dict(color="white", size=13),
                active=3,
                buttons=[
                    dict(method="relayout", label="1Y",
                         args=[{"xaxis.range": [max_date - pd.DateOffset(years=1), max_date],
                                "title": "📊 NVIDIA – Daily Change / Return - Last 1 Year"}]),
                    dict(method="relayout", label="5Y",
                         args=[{"xaxis.range": [max_date - pd.DateOffset(years=5), max_date],
                                "title": "📊 NVIDIA – Daily Change / Return - Last 5 Years"}]),
                    dict(method="relayout", label="10Y",
                         args=[{"xaxis.range": [max_date - pd.DateOffset(years=10), max_date],
                                "title": "📊 NVIDIA – Daily Change / Return - Last 10 Years"}]),
                    dict(method="relayout", label="All",
                         args=[{"xaxis.autorange": True,
                                "title": "📊 NVIDIA – Daily Change / Return Over Time"}])
                ]
            )
        ]
    )

    fig.update_layout(
        title=dict(
            text="📊 NVIDIA – Daily Change / Return in Closing Price",
            font=dict(size=20, color="white"),
            x=0.5
        ),
        xaxis=dict(
            title="Date",
            showgrid=True,
            gridcolor="#34495e",
            showline=True,
            linecolor="#7f8c8d",
            rangeslider=dict(visible=True, thickness=0.05),
            type="date"
        ),
        yaxis=dict(
            title="Daily Change",
            showgrid=True,
            gridcolor="#34495e",
            zeroline=True,
            zerolinewidth=1.5,
            zerolinecolor="#95a5a6"
        ),
        plot_bgcolor="#1e272e",
        paper_bgcolor="#1e272e",
        font=dict(color="white"),
        hovermode="x unified",
        margin=dict(t=110, b=60, l=60, r=60),
        height=650
    )

    fig.show()
```

<p style="font-size: 16px; color: #7f8c8d; padding: 20px; border-radius: 10px;">
    In order to define unusual trading volume, we employ a dynamic threshold based on the statistical distribution of each stock volume's data. Specifically, we define <code>unusual activity</code> as any volume which exceeds <strong>the average volume, plus two standard deviations</strong>:
    <br><br>
    <code style="color: #1abc9c; background-color: #2c3e50; padding: 3px 8px; border-radius: 5px;">threshold = mean + 2 × std</code>
</p>

<p style="font-size: 16px; color: #7f8c8d; padding: 10px 20px; border-radius: 10px;">
    This is based on statistical reasoning: in a normal distribution, approximately 95% of the data is found within two standard deviations of the mean. By choosing this as our threshold, we will identify only the top 5% of extreme values as <strong>unusual</strong>, the true outliers.
</p>

<p style="font-size: 16px; color: #7f8c8d; padding: 10px 20px; border-radius: 10px;">
    Spikes in trading activity generally indicate important market events like earnings releases, geopolitical news or surprise countervailing buyer or seller reactions.  Spikes will allow you to identify periods of heightened interest and possible price changes. For this reason, they are a good indicator for short-term traders as well as long-term analysts.
</p>>

```python
df = df.withColumn("Volume_NVDA", col("Volume_NVDA").cast("double"))
df_pd = df.toPandas()

volume_mean = df_pd["Volume_NVDA"].mean()
volume_std = df_pd["Volume_NVDA"].std()

threshold = volume_mean + 2 * volume_std

df_pd["Unusual_Volume"] = df_pd["Volume_NVDA"] > threshold
```
