<p style="font-size: 18px; color: #f1f1f1; background: #2c3e50; padding: 15px; border-radius: 10px;">
    📥 <strong>Resolve Missing Values and Duplicates:</strong>
</p>

<p style="font-size: 16px; color: #7f8c8d; padding: 20px; border-radius: 10px;">
    <strong><code>Close_NVDA</code></strong>, <strong><code>High_NVDA</code></strong> and <strong><code>Low_NVDA</code></strong>:
</p>
<p style="font-size: 16px; color: #7f8c8d; margin-left: 30px; border-radius: 10px;">
    We use this approach because stock prices follow a <strong>sequential trend</strong>, meaning that each day's price is influenced by the previous day's data. If a trading day's data is missing, the <strong> last known price is the best estimate</strong> until new values appear. This prevents <strong>unrealistic jumps</strong> in stock prices caused by missing data.
    Additionally, since stock trading is <strong>continuous</strong>, gaps in the data can occur because stock exchanges <strong>do not operate on weekends or holidays</strong>. As a result, missing values may appear. Using forward fill (ffill) ensures that the <strong>dataset remains consistent</strong> and prevents artificial dips in stock prices.
    Finally, this method <strong>preserves data consistency for analysis and modelling</strong>, ensuring that financial calculations remain smooth and uninterrupted.
</p>

```python
df[['Close_NVDA', 'High_NVDA', 'Low_NVDA']] = df[['Close_NVDA', 'High_NVDA', 'Low_NVDA']].fillna(method='ffill')
```

<p style="font-size: 16px; color: #7f8c8d; padding: 20px; border-radius: 10px;">
    <strong><code>Market_Cap_NVDA</code></strong> and <strong><code>EV_NVDA</code></strong>
</p>

<p style="font-size: 16px; color: #7f8c8d; margin-left: 30px; border-radius: 10px;">
    For the variables <strong><code>Market_Cap_NVDA</code></strong> and <strong><code>EV_NVDA</code></strong>, we apply the approach of <strong>replacing missing values with zero (0)</strong> because these financial metrics are derived from calculations that may not always be available in certain periods.
</p>

<p style="font-size: 16px; color: #7f8c8d; margin-left: 40px; border-radius: 10px;">
    📌 <strong>Market Capitalization</strong> is calculated as: <code>Market_Cap_NVDA = Close_NVDA × Shares_Out_NVDA</code>
</p>

<p style="font-size: 16px; color: #7f8c8d; margin-left: 40px; border-radius: 10px;">
    📌 <strong>Enterprise Value (EV)</strong> is calculated as: <code>EV_NVDA = Market_Cap_NVDA + Total_Debt_NVDA - Cash_Equivalents_NVDA</code>
</p>

<p style="font-size: 16px; color: #7f8c8d; margin-left: 30px; border-radius: 10px;">
    If any of these components are missing, it means that there is insufficient data to compute a meaningful value. Instead of keeping missing values (<code>NaN</code>), we assign <strong>zero (0)</strong> to prevent issues in further analysis, particularly when performing mathematical operations, where missing values could affect statistical computations and financial models. Additionally, this approach <strong>avoids distortions</strong> in charts and data visualizations, ensuring that calculations remain consistent over time.
</p>

```python
df[['Market_Cap_NVDA', 'EV_NVDA']] = df[['Market_Cap_NVDA', 'EV_NVDA']].fillna(0)
```

<p style="font-size: 16px; color: #7f8c8d; padding: 20px; border-radius: 10px;">
    <strong><code>Total_Debt_NVDA</code></strong> and <strong><code>Cash_Equivalents_NVDA</code></strong>
</p>

<p style="font-size: 16px; color: #7f8c8d; margin-left: 30px; border-radius: 10px;">
    For the variables <strong><code>Total_Debt_NVDA</code></strong> and <strong><code>Cash_Equivalents_NVDA</code></strong>, we handle missing values by filling them with the <strong>median</strong> of the dataset instead of using forward-fill or replacing them with zero.
</p>

<p style="font-size: 16px; color: #7f8c8d; margin-left: 40px; border-radius: 10px;">
    📌 <strong>Total Debt</strong> represents the sum of short-term and long-term liabilities that a company holds.<br>
    📌 <strong>Cash Equivalents</strong> refer to the company's liquid assets that can be quickly converted into cash.
</p>

<p style="font-size: 16px; color: #7f8c8d; margin-left: 30px; border-radius: 10px;">
    Using the median instead of zero helps maintain a more <strong>realistic financial representation</strong> by preventing extreme fluctuations that could result from missing values. Since financial liabilities and cash reserves usually follow a stable trend, the median is a better estimate than the mean, as it reduces the impact of outliers and ensures more <strong>robust statistical imputation</strong>. This approach ensures that missing values do not negatively affect financial calculations such as <code>Enterprise Value (EV_NVDA)</code> and other investment analysis metrics.
</p>

```python
df[['Total_Debt_NVDA', 'Cash_Equivalents_NVDA']] = df[['Total_Debt_NVDA', 'Cash_Equivalents_NVDA']].applymap(lambda x: np.nan if x == 0 else x)
df[['Total_Debt_NVDA', 'Cash_Equivalents_NVDA']] = df[['Total_Debt_NVDA', 'Cash_Equivalents_NVDA']].apply(pd.to_numeric, errors='coerce')

df[['Total_Debt_NVDA', 'Cash_Equivalents_NVDA']] = df[['Total_Debt_NVDA', 'Cash_Equivalents_NVDA']].fillna(df[['Total_Debt_NVDA', 'Cash_Equivalents_NVDA']].median())
```

<p style="font-size: 16px; color: #7f8c8d; padding: 20px; border-radius: 10px;">
    <strong><code>Volume_NVDA</code></strong> and <strong><code>Shares_Out_NVDA</code></strong>
</p>

<p style="font-size: 16px; color: #7f8c8d; margin-left: 30px; border-radius: 10px;">
    Since <code>Volume_NVDA</code> and <code>Shares_Out_NVDA</code> both represent financial data that generally follows a continuous trend (one point comes after another), we apply <strong>linear interpolation</strong> to substitute missed values.  This technique fills in missed values by calculating intermediate values between known data points providing a more seamless and accurate forecast while respecting stock market trends. 
    By invoking <code>interpolate(method='linear')</code> we ensure that missed values are filled with estimates derived from immediately neighbouring data points, in consideration for maintaining the dataset's continuity to avoid any unwanted spikes in the stock data.
</p>

```python
df[['Shares_Out_NVDA', 'Volume_NVDA']] = df[['Shares_Out_NVDA', 'Volume_NVDA']].ffill().bfill().interpolate(method='linear')

# After the resolve of nan values we recalculate Market Cap
df['Market_Cap_NVDA'] = df['Close_NVDA'] * df['Shares_Out_NVDA']

# Recalculate Price-to-Sales Ratio
df['PS_Ratio_NVDA'] = df['Market_Cap_NVDA'] / df['Total_Revenue_NVDA']

# Handle missing or infinite values
df['PS_Ratio_NVDA'] = df['PS_Ratio_NVDA'].replace([np.inf, -np.inf], np.nan)
df['PS_Ratio_NVDA'].fillna(0, inplace=True)
```

<p style="font-size: 16px; color: #7f8c8d; padding: 20px; border-radius: 10px;">
    In order to define unusual trading volume, we employ a dynamic threshold based on the statistical distribution of each stock volume's data. Specifically, we define <code>unusual activity</code> as any volume which exceeds <strong>the average volume, plus two standard deviations</strong>:
    <br><br>
    <code style="color: #1abc9c; background-color: #2c3e50; padding: 3px 8px; border-radius: 5px;">threshold = mean + 2 × std</code>
</p>

<p style="font-size: 16px; color: #7f8c8d; padding: 10px 20px; border-radius: 10px;">
    This is based on statistical reasoning: in a normal distribution, approximately 95% of the data is found within two standard deviations of the mean. By choosing this as our threshold, we will identify only the top 5% of extreme values as <strong>unusual</strong>, the true outliers.
</p>

<p style="font-size: 16px; color: #7f8c8d; padding: 10px 20px; border-radius: 10px;">
    Spikes in trading activity generally indicate important market events like earnings releases, geopolitical news or surprise countervailing buyer or seller reactions.  Spikes will allow you to identify periods of heightened interest and possible price changes. For this reason, they are a good indicator for short-term traders as well as long-term analysts.
</p>>

```python
avg_volume = df['Volume_NVDA'].mean()
std_volume = df['Volume_NVDA'].std()

unusual_threshold = avg_volume + 2 * std_volume
df['Unusual_Activity'] = df['Volume_NVDA'] > unusual_threshold
```
